```python
def file_has_been_logged(filename, logfile=f"{WIKI_FOLDER}/wiki/log.md"):
    if '/' in filename:
        filename = filename.split('/')[-1]
    with open(logfile, "r") as f:
        log_md = f.read()
    log_lines = [line for line in log_md.split("\n") if line.startswith("##")]
    return any(filename in line for line in log_lines)

filename = "data/processed/Argus_Research/2026-11-01/Johnson__Johnson_JNJ_Equity_Research_Report.pdf"
filename = "Johnson__Johnson_JNJ_Equity_Research_Report.pdf"
filename = "This Bond Measure Says Stocks Are Cheap. Don’t Believe It"
file_has_been_logged(filename)
```

```python
""" https://github.com/leafoflegend/obsidian_ingest/tree/master
A small Node.js tool that copies files from a source directory into an Obsidian vault subfolder on a daily schedule. 
It tracks what it has already ingested via a manifest, so re-runs only copy new or changed files.

https://community.obsidian.md/plugins/ingester

data/raw/clippings --> data/processed/clippings/{yyyy-mm-dd --> manual multiple ingest (or automated ingest via the tool)

"""
config_file = f"{WIKI_FOLDER}/.obsidian/plugins/karpathywiki/data.json"
with open(config_file, "r") as f:
    data = f.read()
import json
import datetime
import os
data = json.loads(data)
today_str = datetime.datetime.now().strftime("%Y-%m-%d")
folder = f'data/processed/clippings/{today_str}/'

# create folder if it doesn't exist
os.makedirs(f'{WIKI_FOLDER}/{folder}', exist_ok=True)
data["watchedFolders"] = [folder]
data["watchedFolders"]

with open(config_file, "w") as f:
    json.dump(data, f, indent=2)

print("make sure to restart Obsidian for folder change to take effect")

```
